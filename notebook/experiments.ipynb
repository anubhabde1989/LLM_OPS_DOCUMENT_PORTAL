```python
file_path = os.path.join(os.getcwd(), "data","sample.pdf")
loader = PyPDFLoader(file_path)
loader.load()
```

```python
document=loader.load()
document
```

```python
len(document) #No of pages in the document
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=200, length_function=len) # Experimental parameter values
docs = text_splitter.split_documents(document)
```

```python
test = embedding_model.embed_documents(
    [d.page_content for d in docs]
)

print(len(test), len(docs))
```

```python
embeddings = embedding_model.embed_documents([doc.page_content for doc in docs])

vectorStore = faiss.FAISS.from_documents(docs, embedding_model)
```

```python
relevant_doc=vectorStore.similarity_search("What is a LLM?", k=2)
```

```python
prompt_template = """Answer the question based on the context provided below.
                     If the context does not contain sufficient information, respond with:
                     "I don't have enough information about this"
                     Context: {context}
                     Question: {question}
                     Answer:"""
```

```python
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate(template=prompt_template, input_variables=["context", "question"])
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
```

```python
def format_docs(docs):
    return "\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
llm = model
retriever = vectorStore.as_retriever(search_kwargs={"k": 10})

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain.invoke("Do you know about Anubhab?")
```

```python
rag_chain.invoke("Summarize about transformers in simple terms?")
```
